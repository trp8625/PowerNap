```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader

print(f"PyTorch version: {torch.__version__}")
device = torch.device('mps' if torch.backends.mps.is_available() else 'cpu')
device
```

```python
CONFIG = {
    'window_size': 24,
    'lstm_units': 64,
    'num_layers': 1,
    'dropout': 0.2,
    'epochs': 150,
    'batch_size': 32,
    'learning_rate': 1e-3,
    'patience': 20,
    }
```

```python
df = pd.read_csv("full_dataset_featured.csv", parse_dates=["datetime"])
df = df.sort_values("datetime").reset_index(drop=True)
df.head(10)
```

```python
feature_cols = [
    # weather
    "temp_c", "wind_speed", "cloud_cover", "solar_radiation",
    # lag features
    "carbon_intensity_lag_1", "carbon_intensity_lag_2",
    "carbon_intensity_lag_3", "carbon_intensity_lag_24",
    # cyclical time encodings
    "hour_sin", "hour_cos",
    "day_of_week_sin", "day_of_week_cos",
    "month_sin", "month_cos",
    # energy mix
    "cfe_pct", "re_pct"
]

target_col = 'carbon_intensity'
```

```python
fig, axes = plt.subplots(figsize=(14, 6))

for region, group in df.groupby("region"):
    axes.plot(group["datetime"], group["carbon_intensity"], lw=0.8, label=region)

axes.set_ylabel("Carbon Intensity")
axes.set_xlabel("Datetime")
axes.legend(title="Region", bbox_to_anchor=(1.01, 1), loc="upper left")
plt.suptitle("Carbon Intensity Over Time by Region", fontsize=13)
plt.tight_layout()
plt.show()
```

```python
def create_sequences(X, y, window_size):
    Xs, ys = [], []
    for i in range(len(X) - window_size):
        Xs.append(X[i : i + window_size])    # window_size, n_features
        ys.append(y[i + window_size])        # next step target
    return np.array(Xs), np.array(ys)

def make_splits(group, feature_cols, target_col, window_size):
    group = group.sort_values('datetime').reset_index(drop=True)

    X = group[feature_cols].values       
    y = group[target_col].values.reshape(-1, 1)  

    train_end = int(0.7 * len(group))     
    val_end = int(0.8 * len(group))

    feature_scaler = MinMaxScaler()
    target_scaler = MinMaxScaler()

    X_train = feature_scaler.fit_transform(X[:train_end])
    X_val = feature_scaler.transform(X[train_end:val_end])
    X_test = feature_scaler.transform(X[val_end:])

    y_train = target_scaler.fit_transform(y[:train_end]).flatten().astype(np.float32)
    y_val = target_scaler.transform(y[train_end:val_end]).flatten().astype(np.float32)
    y_test = target_scaler.transform(y[val_end:]).flatten().astype(np.float32)

    X_train_seq, y_train_seq = create_sequences(X_train, y_train, window_size)
    X_val_seq, y_val_seq = create_sequences(X_val, y_val, window_size)
    X_test_seq, y_test_seq = create_sequences(X_test, y_test, window_size)

    return X_train_seq, y_train_seq, X_val_seq, y_val_seq, X_test_seq, y_test_seq, target_scaler
```

```python
class CarbonDataset(Dataset):
    def __init__(self, X, y):
        self.X = torch.tensor(X, dtype=torch.float32)
        self.y = torch.tensor(y, dtype=torch.float32)

    def __len__(self):
        return len(self.X)

    def __getitem__(self, idx):
        return self.X[idx], self.y[idx]
```

```python
class LSTM(nn.Module):
    def __init__(self, n_features, hidden_size, num_layers, dropout):
        super().__init__()
        self.lstm = nn.LSTM(
            input_size=n_features,
            hidden_size=hidden_size,
            num_layers=num_layers,
            dropout=dropout if num_layers > 1 else 0,
            batch_first=True,
        )
        self.dropout = nn.Dropout(dropout)
        self.fc = nn.Linear(hidden_size, 1)

    def forward(self, x):
        out, _ = self.lstm(x)
        out = self.dropout(out[:, -1, :])
        return self.fc(out).squeeze(-1)
```

```python
def train_model(X_train_seq, y_train_seq, X_val_seq, y_val_seq, config, device, save_path):

    train_loader = DataLoader(CarbonDataset(X_train_seq, y_train_seq),
                              batch_size=config['batch_size'], shuffle=False)
    val_loader = DataLoader(CarbonDataset(X_val_seq, y_val_seq),
                              batch_size=config['batch_size'], shuffle=False)

    model = LSTM(X_train_seq.shape[2], config['lstm_units'], config['num_layers'],config['dropout']).to(device)
    criterion = nn.MSELoss()
    optimizer = torch.optim.Adam(model.parameters(), lr=config['learning_rate'])

    train_losses, val_losses = [], []
    best_val_loss    = float('inf')
    patience_counter = 0

    for epoch in range(config['epochs']):
        # Train
        model.train()
        batch_losses = []
        for X_batch, y_batch in train_loader:
            X_batch, y_batch = X_batch.to(device), y_batch.to(device)
            optimizer.zero_grad()
            loss = criterion(model(X_batch), y_batch)
            loss.backward()
            optimizer.step()
            batch_losses.append(loss.item())
        train_loss = np.mean(batch_losses)

        # Validate
        model.eval()
        batch_losses = []
        with torch.no_grad():
            for X_batch, y_batch in val_loader:
                X_batch, y_batch = X_batch.to(device), y_batch.to(device)
                batch_losses.append(criterion(model(X_batch), y_batch).item())
        val_loss = np.mean(batch_losses)

        train_losses.append(train_loss)
        val_losses.append(val_loss)

        # Early stopping
        if val_loss < best_val_loss:
            best_val_loss = val_loss
            torch.save(model.state_dict(), save_path)
            patience_counter = 0
        else:
            patience_counter += 1
            if patience_counter >= config['patience']:
                print(f'Early stopping at epoch {epoch + 1}')
                break

        if (epoch + 1) % 10 == 0:
            print(f'Epoch {epoch+1:3d} | Train: {train_loss:.4f} | Val: {val_loss:.4f}')

    model.load_state_dict(torch.load(save_path))
    print(f'Best val loss: {best_val_loss:.4f}')
    return model, train_losses, val_losses
```

```python
W = CONFIG['window_size']

region_data = {} 
region_models = {}  
region_losses = {}  

for region, group in df.groupby('region'):
    print(f'{region}')

    X_tr, y_tr, X_v, y_v, X_te, y_te, ts = make_splits(
        group, feature_cols, target_col, W
    )
    print(f'train={X_tr.shape} | val={X_v.shape} | test={X_te.shape}')

    region_data[region] = {
        'X_test': X_te, 
        'y_test': y_te,
        'target_scaler': ts,
    }

    model, train_losses, val_losses = train_model(
        X_tr, y_tr, X_v, y_v,
        config = CONFIG,
        device = device,
        save_path = f'best_model_{region}.pt',
    )

    region_models[region] = model
    region_losses[region] = (train_losses, val_losses)
```

```python
regions = list(region_losses.keys())
fig, axes = plt.subplots(2, 2, figsize=(14, 8))

for ax, region in zip(axes.flat, regions):
    train_losses, val_losses = region_losses[region]
    ax.plot(train_losses,label='Train')
    ax.plot(val_losses, label='Val')
    ax.set_title(region.capitalize())
    ax.set_xlabel('Epoch')
    ax.set_ylabel('MSE Loss (scaled)')
    ax.legend()

plt.suptitle('Training History — Per Region', fontsize=13)
plt.tight_layout()
plt.show()
```

```python
results = {}

for region, model in region_models.items():
    X_te = region_data[region]['X_test']
    y_te = region_data[region]['y_test']
    ts = region_data[region]['target_scaler']

    test_loader = DataLoader(CarbonDataset(X_te, y_te), batch_size=CONFIG['batch_size'], shuffle=False)

    model.eval()
    all_preds, all_targets = [], []
    with torch.no_grad():
        for X_batch, y_batch in test_loader:
            preds = model(X_batch.to(device)).cpu().numpy()
            all_preds.append(preds)
            all_targets.append(y_batch.numpy())

    y_pred = ts.inverse_transform(np.concatenate(all_preds).reshape(-1, 1)).flatten()
    y_true = ts.inverse_transform(np.concatenate(all_targets).reshape(-1, 1)).flatten()

    rmse = np.sqrt(mean_squared_error(y_true, y_pred))
    mae = mean_absolute_error(y_true, y_pred)
    r2 = 1 - np.sum((y_true - y_pred)**2) / np.sum((y_true - np.mean(y_true))**2)

    results[region] = {'y_true': y_true, 'y_pred': y_pred, 'rmse': rmse, 'mae': mae, 'r2': r2}
    print(f'{region} | RMSE: {rmse:.4f} | MAE: {mae:.4f} | R^2: {r2:.4f}')

print()
print(f"Overall RMSE: {np.mean([v['rmse'] for v in results.values()]):.4f} gCO2/kWh")
print(f"Overall MAE: {np.mean([v['mae'] for v in results.values()]):.4f} gCO2/kWh")
print(f"Overall R^2: {np.mean([v['r2'] for v in results.values()]):.4f}")
```
